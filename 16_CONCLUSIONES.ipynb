## 1. Serie Temporal

### ¿Cual es el modelo mas adecuado?

Una vez realizados varios modelos para predecir la energía producida por el inversor, se ha llegado a la conclusión de que los 3 modelos utilizados (Regresión lineal, Random Forest y XGBoost) predicen con un error bastante elevado. Dentro de estos tres, el que menos error tiene es el XGBoot, pero, ¿Porque ocurre esto?

- El modelo XGBoost realiza un ajuste iterativo y corrige errores, esto permite que cada nuevo árbol corrija los errores de los árboles anteriores, esto mejora el modelo de manera progresiva, lo que le da ventaja frente al Random Forest que crea árboles de manera independiente.
- XGBooost es muy eficiente en la captura de interacciones entre características, al tener varias relaciones no lineales entre las variables, este modelo puede identificar y aprovechar mejor estas relaciones que la regresión lineal o el Random Forest.

Debido a las dificultades para la predicción de la variables energía mediante una serie temporal, se descompone la variable Fecha_Hora en multiples columnas y se convierte en una regresión.
Con este nuevo dataframe se van a predecir tanto la energía como la temperatura ambiente y la temperatura interna del inversor.

## 2. Red Neuronal Energía

Para la predicción de la energía se toma como primer modelo una regresión lineal, algo sencillo para comenzar, una vez entrenado y evaluado da como resultado un MAE de 1843, por lo que se decide entrenar una red neuronal para ver si es capaz de bajar este parametro.

Se grafican tanto la curva de pérdida como la curva de MAE, obteniendo las siguientes conclusiones:


1. Épocas iniciales (0 a 10): ambas curvas disminuyen juntas, esto es señal de que el modelo está aprendiendo bien tanto los datos de entrenamiento como los de validación.

2. A partir de la época 10 en adelante: la curva de validación empieza a tener altibajos, esto puede deberse a un sobreajuste (overfitting) o que los datos de validación son muy variables, la curva de entrenamiento sigue bajando de forma suave y estable, lo que indica que es bueno para el set de entrenamiento.

Se realiza una clusterización pero los modelos entrenados para cada cluster no suponen ninguna mejora.





## 3. Temperatura ambiente.

Para la temperatura ambiente también se empieza por el modelo de regresión lineal, que nos da mejores resultados en los datos de validación que en los datos de test.

Se procede a probar el modelo Random Forest para mejorar los datos en el test, una vez entrenado el modelo con los datos de validación se le pasan los datos del test y estos mejoran ostensiblemente a la regresión lineal, en torno a un 40% mas de acierto. Por lo que se considera que este es el mejor modelo para predecir la temperatura ambiente.

Paralelamente se realiza una clusterización para comprobar si se puede mejorar el modelo, pero viendo el resultado del Random Forest en los datos de test se desestima esta opción pensando en el tiempo que tiener que invertir para entrenar varios modelos para cada cluster, por lo que la mejora que se pudiera obtener y que no se está calculando se asume como pérdida.

## 4. Temperatura interna del inversor

Para la temperatura interna del inversor se realiza el mismo procedimiento que para la temperatura ambiente, también nos da mejores resultados en la validación que en el test, por lo que se prueba con el modelo Random Forest y con el XGBoost, como estos siguen arrojando bastante error en el test, se genera una red neuronal para intentar mejorar los resultados.
La red neuronal entrenada no mejora los resultados del XGBoost, por lo que elegiremos este modelo para la predicción de la temperatura interna de los inversores.
