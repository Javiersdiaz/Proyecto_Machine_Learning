En este documento se van a plantear diferentes variables objetivos que se consideran de importancia para el correcto funcionamiento de la planta y para la seguridad de los trabajadores in situ.

Para ello se van a estudiar tres variables:

- Energía.
- Temperatura ambiente.
- Temperatura interna del inversor.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df = pd.read_csv(r"C:\Users\javier.sdiaz\Proyecto ML_2025\data\processed\serie_temporal_procesada.csv", sep=",")
df['Fecha_Hora'] = pd.to_datetime(df['Fecha_Hora'])
```

# 1. Energía

Lo primero que hay que saber es que un parque fotovoltaico es mas rentable cuanta mas energía produce, esto genera mas ingresos y es un buen indicador de que la instalación está funcionando correctamente, por lo que se va a predecir la producción de energía basandonos en los diferentes parametros que tenemos, también puede ser util para si un inversor falla poder calcular la pérdida de energía que se ha tenido en ese periodo y por tanto el coste económico.

```python
# Crear gráfico interactivo
fig = px.line(df, x='Fecha_Hora', y='Energia', title='Evolución de energía a lo largo del tiempo')

# Personalización adicional (opcional)
fig.update_layout(
    xaxis_title='Fecha y Hora',
    yaxis_title='Energia',
    xaxis=dict(showgrid=True),
    yaxis=dict(showgrid=True)
)

fig.show()
```

# 2. Temperatura ambiente

El clima en esta zona puede variar durante la misma semana por lo que sobre todo en los meses de verano hay que extremar las precauciones con los operarios, la ley estipula que no se puede trabajar en una planta fotovoltaica por encima de los 30ºC de temperatura, por lo que una de las variables a estudiar será la temperatura durante el día

```python
# Calcular el porcentaje
total = len(df)
mayores_30 = df[df['Temp_amb'] > 30].shape[0]
porcentaje = (mayores_30 / total) * 100

# Crear una nueva columna para la condición de color
df['color'] = df['Temp_amb'].apply(lambda x: 'Temperatura > 30' if x > 30 else 'Temperatura ≤ 30')

import plotly.graph_objects as go

# Crear una máscara
mask_high = df['Temp_amb'] > 30
mask_low = df['Temp_amb'] <= 30

fig = go.Figure()

# Línea azul continua para todo (Temperatura ≤ 30)
fig.add_trace(go.Scatter(
    x=df.loc[mask_low, 'Fecha_Hora'],
    y=df.loc[mask_low, 'Temp_amb'],
    mode='lines',
    name='Temperatura ≤ 30°C',
    line=dict(color='blue', width=2)
))

# Línea roja continua para Temp > 30°C
fig.add_trace(go.Scatter(
    x=df.loc[mask_high, 'Fecha_Hora'],
    y=df.loc[mask_high, 'Temp_amb'],
    mode='lines',
    name='Temperatura > 30°C',
    line=dict(color='red', width=2)
))

# Anotación con porcentaje
fig.add_annotation(
    text=f"{porcentaje:.2f}% de los periodos con Temp > 30°C",
    xref="paper", yref="paper",
    x=0.5, y=1.12,
    showarrow=False,
    font=dict(size=14, color="black"),
    bgcolor="rgba(255,255,255,0.8)",
    bordercolor="black",
    borderwidth=1,
    borderpad=4
)

fig.update_layout(
    title='Evolución Temperatura ambiente',
    xaxis_title='Fecha y Hora',
    yaxis_title='Temp_amb',
    template='plotly_white',
    xaxis=dict(showgrid=True),
    yaxis=dict(showgrid=True),
    margin=dict(t=100)
)

fig.show()



```

# 3. Temperatura interna del inversor

Es necesario estudiar esta variable ya que la mayor parte de las veces en las que el inversor falla es por un aumento en su temperatura interna, esto ocurre normalmente en los meses de verano en los que la radiación y temperatura son mas altas, por lo que cuando la temperatura interna es mayor a 40ºC hay que tomar medidas para bajar esta temperatura o simplemente apagar el inversor para que no haya accidentes.

```python
# Calcular el porcentaje
total = len(df)
mayores_40 = df[df['Temp_int'] > 40].shape[0]
porcentaje = (mayores_40 / total) * 100

# Crear una nueva columna para la condición de color
df['color_2'] = df['Temp_int'].apply(lambda x: 'Temperatura > 40' if x > 30 else 'Temperatura ≤ 40')

import plotly.graph_objects as go

# Crear una máscara
mask_high = df['Temp_int'] > 40
mask_low = df['Temp_int'] <= 40

fig = go.Figure()

# Línea azul continua para todo (Temperatura_int ≤ 40)
fig.add_trace(go.Scatter(
    x=df.loc[mask_low, 'Fecha_Hora'],
    y=df.loc[mask_low, 'Temp_int'],
    mode='lines',
    name='Temperatura ≤ 40°C',
    line=dict(color='blue', width=2)
))

# Línea roja continua para Temp > 40°C
fig.add_trace(go.Scatter(
    x=df.loc[mask_high, 'Fecha_Hora'],
    y=df.loc[mask_high, 'Temp_int'],
    mode='lines',
    name='Temperatura > 40°C',
    line=dict(color='red', width=2)
))

# Anotación con porcentaje
fig.add_annotation(
    text=f"{porcentaje:.2f}% de los periodos con Temp > 40°C",
    xref="paper", yref="paper",
    x=0.5, y=1.12,
    showarrow=False,
    font=dict(size=14, color="black"),
    bgcolor="rgba(255,255,255,0.8)",
    bordercolor="black",
    borderwidth=1,
    borderpad=4
)

fig.update_layout(
    title='Evolución Temperatura interna del inversor',
    xaxis_title='Fecha y Hora',
    yaxis_title='Temp_int',
    template='plotly_white',
    xaxis=dict(showgrid=True),
    yaxis=dict(showgrid=True),
    margin=dict(t=100)
)

fig.show()
```
