Buena pregunta. Que la regresión lineal te salga como el mejor modelo en una serie temporal puede deberse a varias razones. Aquí te comparto algunas posibles explicaciones:

1. Tendencia dominante
Si tu serie temporal tiene una tendencia clara (creciente o decreciente) y no tiene estacionalidad ni mucha variabilidad no explicada, la regresión lineal captura bien esa tendencia.
En estos casos, modelos más complejos (ARIMA, Prophet, redes neuronales, etc.) pueden no aportar una mejora significativa y la regresión lineal es suficiente.
2. Poca variabilidad/noise
Si la serie es suave y tiene poco ruido o fluctuaciones aleatorias, un modelo sencillo como la regresión lineal podría tener buen rendimiento en términos de error (MSE, MAE, etc.).
3. Horizonte de predicción corto
Si estás evaluando la precisión en un horizonte de predicción muy corto (por ejemplo, predecir solo el siguiente paso o pocos pasos adelante), la regresión lineal a veces es difícil de superar, especialmente si los datos son estables.
4. Datos insuficientes
A veces, si tienes pocos datos o la serie es muy corta, los modelos más sofisticados no logran generalizar bien y tienden a sobreajustar. En cambio, la regresión lineal, al tener pocos parámetros, funciona mejor.
5. Mala configuración de otros modelos
Puede ser que los otros modelos (ARIMA, LSTM, Prophet, etc.) no estén bien tuneados (malos hiperparámetros, no están capturando estacionalidad, etc.), y por eso la regresión lineal gana en la comparación.
6. La métrica elegida
Dependiendo de la métrica que uses para comparar (RMSE, MAE, MAPE, etc.), es posible que la regresión lineal sea la mejor según esa métrica, aunque otros modelos podrían ser mejores para otros objetivos (capturar patrones estacionales, picos, etc.).
